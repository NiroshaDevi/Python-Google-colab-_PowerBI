### Result (Handling Missing Values):
**No Missing values found in the DataSet.**

since marks allocated for handling missing values codes are written below:

```python
# Handle missing values (impute or drop)

# filling categorical with mode, numerical with median (or drop if few)
for col in df.columns:
    if df[col].dtype == 'object':
        df[col].fillna(df[col].mode()[0],inplace = True)
    else:
        df[col].fillna(df[col].mode()[0],inplace = True)

# Verifying again
df.isnull().sum()
```

## Result (Handling Duplicates):
**No Duplicates found in the DataSet.**

since marks allocated for handling Duplicates, codes are written below:

```python
# Handle duplicates

# Check duplicates
df.duplicated().sum()

# Drop duplicates
df.drop_duplicates(inplace=True)

# Verifying again
df.duplicated().sum()
```

```python
# Treat outliers if required

# Identify numerical columns
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

# Boxplots to visualize outliers
for col in numeric_cols:
    plt.figure(figsize=(12,1))
    sns.boxplot(x=df[col])
    plt.title(f'Outlier Detection for {col}')
    plt.show()
```

### Result (Handling Outliers):
**No outliers found for the numerical columns in the Dataset.**

```python
# Check skewness and apply transformations

# Check skewness
df[numeric_cols].skew()
```

## Result(checking Skewness):
**All numeric columns have skewness values between -0.01 and +0.02,
The data is approximately normally distributed.**

No transformations (like log or sqrt) are required for these variables.

Since marks are allocated, codes written below:

```python
# Apply log transformation
for col in numeric_cols:
    if abs(df[col].skew()) > 1:   # strong skew
        df[col] = np.log1p(df[col])  # log(1+x) transformation

# Recheck skewness
df[numeric_cols].skew()
```

```python
# Convert data types if needed

# Convert columns with Yes/No to boolean
df['Discount Applied'] = df['Discount Applied'].map({'Yes': 1, 'No': 0})
df['Promo Code Used'] = df['Promo Code Used'].map({'Yes': 1, 'No': 0})
df['Subscription Status'] = df['Subscription Status'].map({'Yes': 1, 'No': 0})
```

```python
#checking the Changes
display(df.iloc[1057:1101])
```

```python
df.dtypes
```

## Result (Convert DataTypes):
Converted categorical fields with Yes/No values to binary (0/1) and ensured their datatypes are in appropriate format.

```python
# Feature transformations (date parts, derived fields if required for analysis)

#Creating Age Group
bins = [0, 18, 25, 35, 50, 65, 100]
labels = ['Teen', 'Young Adult', 'Adult', 'Middle Age', 'Senior', 'Elder']
df['Age Group'] = pd.cut(df['Age'], bins=bins, labels=labels, right=False)

# Average Purchase per previous purchase ratio
df['Avg_Spend_per_Purchase(USD)'] = df['Purchase Amount (USD)'] / (df['Previous Purchases'] + 1)

# Verify new features
df[['Age', 'Age Group', 'Purchase Amount (USD)', 'Previous Purchases', 'Avg_Spend_per_Purchase(USD)']].head()
```

* The smaller slices indicate seasons with lower transaction volumes.

* The exact percentages displayed on the chart show how purchases are distributed throughout the year.

**What is it saying?**

* Winter records the highest proportion of purchases, suggesting strong customer demand during the colder months — possibly due to holidays and seasonal shopping.

* Summer and Spring show moderate levels of activity, reflecting stable but lower demand periods.

* Fall has the fewest purchases, indicating it’s the slowest retail season in this dataset.

* The pattern indicates that seasonal factors significantly influence shopping behavior.

**What features you used?**

* Season->Categorical->Represents the time of year when the purchase occurred — Winter, Spring, Summer, or Fall. Used to group and compare customer transactions across seasons.

* Customer ID / Transaction Count->Count (derived)->Each customer record or transaction contributes to the total count of purchases per season. Used to measure how many purchases happened in each season.

**From this what you are showing us?**
